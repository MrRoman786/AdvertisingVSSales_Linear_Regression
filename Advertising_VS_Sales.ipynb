```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
data={
    "Advertisment":[120,130,135,150,160,163,168,173,180,199],
    "Sales":[21000,24000,26000,30000,35000,38000,42000,46000,50000,55000]
}
data_set=pd.DataFrame(data)
x=data_set[["Advertisment"]]
y=data_set["Sales"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.3,random_state=42)
lr=LinearRegression()
lr.fit(x_train,y_train)
prediction=lr.predict([[181]])
print(f"Predicted Value for {181} is : {round(prediction[0],2)}")
```
